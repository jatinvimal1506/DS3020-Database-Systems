```python
import pymongo
import json 
```

```python
with open("employees_dataset.json") as f:
    records = json.load(f)
```

```python
#Q1 a
client = pymongo.MongoClient()
company_db = client["newdatabase"]

employee = company_db["employees"]
```

```python
#Q1 b
employee.insert_many(records)
```

```python
#Q2
for doc in employee.find(limit=5):
    print(doc)
```

```python
#Q3 
print(employee.count_documents({}))  #Command to print number of doucments
print(employee.count_documents({"dept":"HR"})) # Find the number of employees with dept HR
```

```python
#Q4 a
results = employee.update_one({"name": "Nisha Patel"}, {"$set": {"salary":50000}})

for doc in employee.find({"name": "Nisha Patel"}):
    print(doc)

print(f"Matched Count : {results.matched_count}")
print(f"Matched Count : {results.modified_count}")
```

```python
#Q4 b
result = employee.delete_one({"name": "Kiran Swamy"})
print(f"Deleted Count : {result.deleted_count}")
```

```python
#Q5 a
for doc in employee.find({'salary': {'$gt': 100000}}, ["name","salary"]).sort("salary",-1):
    del doc["_id"]
    print(doc)
```

```python
#Q5 b
for doc in employee.find({'$or': [{'dept': "IT"},{"salary":{"$gt":120000}}]},["name","dept","salary"]):
    del doc["_id"]
    print(doc)


```

```python
#Q6

result = employee.update_many({"dept":"HR"},{"$inc":{"salary":15000}})
print(result)

result = employee.update_many({"dept":"HR"},{"$push":{"skills":"Leadership"}})
```

```python
#Q7
for doc in employee.find({"skills": {"$elemMatch": {"$gte": "Java", "$lt": "SQL"}}},["name","skills"]):
    del doc["_id"]
    print(doc)


```

```python
#Q8
pipeline = [
    {"$unwind":"$skills"},
    {"$group":{"_id":"$skills","employee_count":{"$sum":1}}},
    {"$sort":{"employee_count":-1}}
]
cnt = 0
result = employee.aggregate(pipeline)
for doc in result:
    if cnt == 3:
        break

    print(doc)
    cnt+=1

    
```

```python
#Q9
for doc in employee.find({"$nor": [{"dept": "Finance"}, {"salary": {"$gt": 100000}},{"address.city": "Mumbai"}]},["name","dept","address.city","salary"]):
    del doc["_id"]
    print(doc)
```

```python
#Q10
for doc in employee.find({'name': {"$regex": "^S"},'address.city': {"$regex": "i$"}}, ['name',"address.city"]):
    del doc["_id"]
    print(doc)
```
